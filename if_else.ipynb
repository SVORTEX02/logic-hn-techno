```python
cost=input("enter  price of bike: ")

if cost.isdigit() and int(cost)>0:
    cost=int(cost)
    tax=0
    if cost>10000:
        tax=15
    elif cost>5000 and cost<=10000:
        tax=10
    elif cost<=5000:
        tax=5
    
    tax_amount = (tax * cost) / 100
    total_cost = cost + tax_amount

    print(f"tax Rate{tax}")
    print(f"road tax amount: {tax_amount}")
    print(f"total price after tax: {total_cost}")
else:
    print("enter a valid cost")
    
```

```python
# leap year
year=input("enter a year").strip()
if year.isdigit():
    year=int(year)
    if year%4==0 or (year%100!=0 or year%400==0 ):
        print(f"{year} is a leap year")
    else:
        print(f"{year} is not a leap year")
```
